```python

#1. Reverse a string
def revString(s)->str:
  return s[::-1]
s="qwerty"
revString(s)
```

```python
# 2.Check if string is palindrome
def check_peli(s):
  return s==s[::-1]
s1="madam"
s2="qwerty"
check_peli(s1),check_peli(s2)
```

```python
# 3.Count vowels in a string

def countVowels(s):
  vowels="aeiouAEIOU"
  count=0
  for ch in s:
    if ch in vowels:
      count+=1
  return count
s="qweriiiiuTTEEAAAIty"
countVowels(s)

```

```python
# 4.Count characters in a string
def countString(s):
  count=0
  for ch in s:
    if ch.isalpha():
      count+=1
  return count
s="qweriiiiuTTEEAAAIty"
countString(s)
```

```python
# 5.Count consonants in a string
def countConsonants(s):
  consonants="bcdfghjklmnpqrstvwxyzBCDFGHJKLMNPQRSTVWXYZ"
  count=0
  for ch in s:
    if ch in consonants:
      count+=1
  return count
a="qwertgcswertghbvcdertgvcaaaaaaaaaaaaaaaeeeeeee"
print(len(a))
countConsonants(a)
```

```python
# 6.Count words in a string
def countWords(s):
  words=s.split()
  return len(words)
s="qwertgcswe rtghbvcdertgvcaa aaaaaaaaaaaa aeeeee ee 123"
countWords(s)
```

```python
# 7.Convert string to lowercase
# 8.Convert string to uppercase
def convert(s,choice):
  if choice=="up"or choice=="UP":
    return s.upper()
  elif choice=="lower" or choice=="LOWER":
    return s.lower()
s="qwertgcswe"
convert(s,"lower")
convert(s,"up")

```

```python

# 9.Swap first and last characters
def swap(s):
  result=s[-1]+s[1:-1]+s[0]
  return result
s="qwerty"
swap(s)
```

```python

# 10.Remove spaces from a string
s="qwerty qwertyhvds werth vwertyujnb vcwerthjnbvc"
def removeSpaces(s):
  return s.replace(" ","")
removeSpaces(s)
```

```python
# 11.Replace spaces with underscore
s="qwerty qwertyhvds werth vwertyujnb vcwerthjnbvc"
def replce(s):
  return s.replace(" ","_")
replce(s)
```

```python
# 19.Count uppercase and lowercase letters
def a(d):
  upper=0
  lower=0
  for ch in d:
    if ch.isupper():
      upper+=1
    elif ch.islower():
      lower+=1
  return upper,lower
s="QWeRDSWEdsweFDSQWERfRTFD"
a(s)
```

```python
# 24.Remove special characters
def removeSpecial(s):
  count=0
  for ch in s:
    if not ch.isalpha():
      if not ch.isdigit():
        count+=1
  return count
s="qwertgfd345567tr@@##@$"
removeSpecial(s)
```
